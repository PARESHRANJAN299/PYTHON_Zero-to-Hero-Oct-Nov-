# Calculate karo:
# 1. total_rows = ?
# 2. total_bytes = ?
# 3. total_gb = ?  (1 GB = 1073741824 bytes)
# 4. kya total_rows 25 million se zyada hai? (True/False)

```python
daily_rows = 40000000
days = 7 
row_size_bytes = 250


#1. total rows 
total_rows = daily_rows * days
#print(total_rows)
## 2. total_bytes = ?

total_bytes = total_rows // row_size_bytes
#print(total_bytes)

#3.
total_gb1 = 1073741824
total_gb = total_bytes // total_gb1
print(total_gb)


#4. # 4. kya total_rows 25 million se zyada hai? (True/False)

rows =  total_rows 
total = 2500000

if rows == total:
    print("True")
    
else: print ("False")

```
